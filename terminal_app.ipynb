```python
# sms_spam_app.py
import joblib
import string
from nltk.corpus import stopwords
import nltk

# Download stopwords if not already done
nltk.download('stopwords')

# -------------------------------
# 1. Load saved models & transformers
# -------------------------------
path = "./models"
spam_detect_model = joblib.load(f'{path}/spam_detect_model.pkl')
count_vectorizer = joblib.load(f'{path}/count_vectorizer.pkl')
tfidf_transformer = joblib.load(f'{path}/tfidf_transformer.pkl')
ig_selector = joblib.load(f'{path}/ig_selector.pkl')

# -------------------------------
# 2. Preprocessing function
# -------------------------------
def text_preprocess(message):
    # Remove punctuation
    nopunc = ''.join([c for c in message if c not in string.punctuation]).lower()
    # Remove stopwords and non-alphabetic tokens
    return ' '.join([w for w in nopunc.split() if w.isalpha() and w not in stopwords.words('english')])

# def text_preprocess(mess):
#     nopunc = [char for char in mess if char not in string.punctuation]

#     # Join the characters again to form the string.
#     nopunc = ''.join(nopunc)
#     nopunc = nopunc.lower()
    
#     # Now just remove any stopwords and non alphabets
#     nostop=[word for word in nopunc.split() if word.lower() not in stopwords.words('english') and word.isalpha()]
    
#     return nostop

# -------------------------------
# 3. Prediction function
# -------------------------------
def predict_message(message):
    # Preprocess
    msg_processed = text_preprocess(message)
    
    # Convert to BoW
    msg_bow = count_vectorizer.transform([msg_processed])
    
    # Convert to TF-IDF
    msg_tfidf = tfidf_transformer.transform(msg_bow)
    
    # Apply Information Gain feature selection
    msg_ig = ig_selector.transform(msg_tfidf)
    
    # Predict label
    prediction = spam_detect_model.predict(msg_ig)[0]
    return prediction

# -------------------------------
# 4. CLI loop for new messages
# -------------------------------
if __name__ == "__main__":
    print("SMS Spam Detection App")
    print("Type 'exit' to quit.\n")
    
    while True:
        message = input("Enter a new SMS message: ")
        if message.lower() == 'exit':
            print("Exiting app.")
            break
        
        label = predict_message(message)
        print(f"Predicted label: {label}\n")
```
